## O que é o Pandas?

O **Pandas** é a biblioteca mais importante para *análise e manipulação de dados* em Python.
Ela introduz duas estruturas de dados fundamentais:

- **`Series`**: um vetor unidimensional rotulado (como uma coluna).
- **`DataFrame`**: uma tabela bidimensional (linhas e colunas), semelhante a uma planilha
  do Excel ou a uma tabela de banco de dados.

### Por que o Pandas importa para Ciência de Dados?

explorar e transformar** os dados. Estima-se que cientistas de dados passem cerca de
**80% do tempo** nessa etapa de preparação — e o Pandas é a ferramenta central para isso.

## 1. Importando as bibliotecas

Começamos importando o `pandas` (convencionalmente apelidado de `pd`), o `numpy`
(para operações numéricas) e o `seaborn` (que traz *datasets* públicos prontos para uso).

```python
# Import core libraries for data manipulation and analysis
import pandas as pd
import numpy as np
import seaborn as sns

# Display all columns when printing DataFrames
pd.set_option("display.max_columns", None)

print("pandas version:", pd.__version__)
print("numpy version:", np.__version__)
print("seaborn version:", sns.__version__)
```

## 2. Criando DataFrames do zero

Existem várias formas de criar um `DataFrame`. As duas mais comuns são:

1. A partir de um **dicionário** (as chaves viram nomes de colunas).
2. A partir de uma **lista de dicionários** (cada dicionário vira uma linha).

```python
# Method 1: create a DataFrame from a dictionary
# Each key becomes a column, each list becomes the column values
students = {
    "name": ["Ana", "Bruno", "Carla", "Diego"],
    "age": [23, 31, 27, 45],
    "city": ["Palmas", "Recife", "Palmas", "Salvador"],
}
df_students = pd.DataFrame(students)
df_students
```

```python
# Method 2: create a DataFrame from a list of dictionaries
# Each dictionary becomes a row
records = [
    {"name": "Elena", "age": 29, "city": "Fortaleza"},
    {"name": "Felipe", "age": 38, "city": "Palmas"},
]
df_records = pd.DataFrame(records)
df_records
```

## 3. Carregando datasets do seaborn (iris e tips)

O `seaborn` disponibiliza *datasets* públicos e reais, muito usados para aprendizado:

- **`iris`**: medidas de flores (comprimento/largura de sépalas e pétalas) de 3 espécies.
- **`tips`**: registros de gorjetas em um restaurante (valor da conta, gorjeta, dia, etc.).

Usamos `sns.load_dataset()` para carregá-los diretamente como `DataFrame`.

```python
# Load two real public datasets bundled with seaborn
iris = sns.load_dataset("iris")
tips = sns.load_dataset("tips")

print("iris shape:", iris.shape)
print("tips shape:", tips.shape)
```

## 4. Inspeção básica dos dados

Antes de analisar, sempre inspecionamos o *dataset* para entender sua estrutura.
Os métodos e atributos mais usados são:

| Método/Atributo | O que faz |
|-----------------|-----------|
| `.shape`        | Retorna (nº de linhas, nº de colunas) |
| `.dtypes`       | Mostra o tipo de dado de cada coluna |
| `.head(n)`      | Exibe as primeiras `n` linhas (padrão 5) |
| `.tail(n)`      | Exibe as últimas `n` linhas |
| `.info()`       | Resumo: colunas, valores não-nulos e tipos |
| `.describe()`   | Estatísticas descritivas das colunas numéricas |

```python
# .shape -> (rows, columns)
print("iris shape:", iris.shape)

# .dtypes -> data type of each column
print("\niris dtypes:")
print(iris.dtypes)
```

```python
# .head() shows the first rows (default = 5)
iris.head()
```

```python
# .tail() shows the last rows
iris.tail(3)
```

```python
# .info() gives a concise technical summary of the DataFrame
iris.info()
```

```python
# .describe() computes descriptive statistics for numeric columns
iris.describe()
```

## 5. Seleção de colunas

Podemos selecionar:

- **Uma coluna** → retorna uma `Series`: `df["coluna"]`
- **Múltiplas colunas** → retorna um `DataFrame`: `df[["col1", "col2"]]` (note a lista dupla `[[ ]]`)

```python
# Select a single column -> returns a Series
species = iris["species"]
print(type(species))
species.head()
```

```python
# Select multiple columns -> returns a DataFrame (note the double brackets)
subset = iris[["sepal_length", "species"]]
subset.head()
```

## 6. Filtragem de linhas com condições booleanas

Podemos filtrar linhas usando **condições booleanas**. A condição gera uma máscara
(uma `Series` de `True`/`False`) que seleciona apenas as linhas desejadas.

Para combinar condições use `&` (E), `|` (OU) e envolva cada condição em parênteses.

```python
# Filter: only rows where species is "setosa"
setosa = iris[iris["species"] == "setosa"]
print("Number of setosa rows:", setosa.shape[0])
setosa.head()
```

```python
# Combine multiple conditions with & (and) / | (or)
# Long-petaled virginica flowers
big_virginica = iris[(iris["species"] == "virginica") & (iris["petal_length"] > 6.0)]
big_virginica
```

## 7. Ordenação com `.sort_values()`

O método `.sort_values()` ordena o `DataFrame` por uma ou mais colunas.
Use `ascending=False` para ordem decrescente.

```python
# Sort the tips dataset by total_bill in descending order
tips_sorted = tips.sort_values("total_bill", ascending=False)
tips_sorted.head()
```

```python
# Sort by multiple columns: first by day, then by tip (descending)
tips.sort_values(["day", "tip"], ascending=[True, False]).head()
```

## 8. Agregação básica com `.groupby()`

O `.groupby()` agrupa linhas que compartilham um valor em comum e permite calcular
estatísticas por grupo (média, contagem, soma, etc.). É a base da análise exploratória.

```python
# Average tip per day of the week
avg_tip_by_day = tips.groupby("day", observed=True)["tip"].mean()
avg_tip_by_day
```

```python
# Count number of records per species in the iris dataset
count_by_species = iris.groupby("species", observed=True)["sepal_length"].count()
count_by_species
```

```python
# Multiple aggregations at once with .agg()
tips.groupby("day", observed=True)["total_bill"].agg(["mean", "count", "max"])
```

## 9. Tratamento de valores ausentes (missing values)

Dados reais quase sempre têm valores ausentes (`NaN`). Os principais métodos são:

- `.isnull()` → indica quais células estão vazias (`True`/`False`).
- `.dropna()` → remove linhas (ou colunas) com valores ausentes.
- `.fillna(valor)` → preenche valores ausentes com um valor definido.

Vamos criar um pequeno `DataFrame` com valores ausentes para demonstrar.

```python
# Create a small DataFrame with some missing values (np.nan)
data_missing = pd.DataFrame({
    "student": ["Ana", "Bruno", "Carla", "Diego"],
    "score": [8.5, np.nan, 7.0, np.nan],
    "hours_studied": [10, 5, np.nan, 8],
})
data_missing
```

```python
# .isnull() shows which cells are missing; .sum() counts them per column
data_missing.isnull().sum()
```

```python
# .dropna() removes rows that contain any missing value
data_missing.dropna()
```

```python
# .fillna() fills missing values -- here we fill scores with the column mean
filled = data_missing.copy()
filled["score"] = filled["score"].fillna(filled["score"].mean())
filled["hours_studied"] = filled["hours_studied"].fillna(0)
filled
```

## 10. Aplicação à Linguística e ao Ensino de Inglês 🎓

Como venho da **Linguística/Sociolinguística** e tenho interesse em **PLN aplicado à educação**,
vou criar um pequeno `DataFrame` **pedagógico** com palavras do vocabulário de inglês e suas
frequências.

> ⚠️ **Nota:** os números abaixo são um **exemplo didático criado manualmente** apenas para
> ilustrar as operações do Pandas. **Não** representam dados reais de frequência de um *corpus*.
> Em projetos reais, essas contagens viriam de *corpora* como o COCA ou o BNC.

Esse tipo de tabela (palavra → frequência) é exatamente a estrutura usada em muitas tarefas de
PLN, como a construção de listas de vocabulário graduadas por nível para aprendizes de inglês.

```python
# Pedagogical example (manually created for teaching purposes -- NOT real corpus data)
# A small vocabulary list with illustrative frequency counts and CEFR levels
vocab = pd.DataFrame({
    "word": ["the", "school", "teacher", "linguistics", "corpus", "morpheme"],
    "part_of_speech": ["article", "noun", "noun", "noun", "noun", "noun"],
    "frequency": [1200, 340, 180, 25, 12, 4],   # illustrative counts only
    "cefr_level": ["A1", "A1", "A2", "C1", "C2", "C2"],
})
vocab
```

```python
# We can apply everything learned today to this vocabulary DataFrame:

# 1) Filter: advanced vocabulary (C1/C2 level)
advanced = vocab[vocab["cefr_level"].isin(["C1", "C2"])]
print("Advanced vocabulary (C1/C2):")
print(advanced[["word", "frequency"]].to_string(index=False))

# 2) Sort by frequency (most frequent first)
print("\nSorted by frequency:")
print(vocab.sort_values("frequency", ascending=False)[["word", "frequency"]].to_string(index=False))
```

```python
# 3) Aggregate: average frequency per CEFR level
avg_freq_by_level = vocab.groupby("cefr_level", observed=True)["frequency"].mean()
print("Average (illustrative) frequency per CEFR level:")
avg_freq_by_level
```
